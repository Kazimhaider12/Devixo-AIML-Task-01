### Part 1: Data Loading & Understanding

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set display settings
pd.set_option('display.max_columns', None)
sns.set_theme(style="whitegrid")

# Load raw dataset
df = pd.read_csv('../dataset/employee_data.csv')
print('Dataset Shape:', df.shape)
df.head()
```

```python
# Display dataset overview
df.info()
df.describe(include='all')
```

### Part 2: Data Cleaning & Preprocessing

```python
# 1. Check Duplicates and Missing Values
print('Duplicate rows:', df.duplicated().sum())
print('\nMissing values per column:\n', df.isnull().sum())

# 2. Remove Duplicates
cleaned_df = df.drop_duplicates().copy()

# 3. Handle Missing Values via Logical Imputation
cleaned_df['Age'].fillna(cleaned_df['Age'].median(), inplace=True)
cleaned_df['Years_Experience'].fillna(cleaned_df['Years_Experience'].median(), inplace=True)
cleaned_df['Monthly_Salary'] = cleaned_df.groupby('Department')['Monthly_Salary'].transform(lambda x: x.fillna(x.median()))

# 4. Type Casting
cleaned_df['Age'] = cleaned_df['Age'].astype(int)
cleaned_df['Years_Experience'] = cleaned_df['Years_Experience'].astype(int)

print('\nCleaned Dataset Shape:', cleaned_df.shape)
print('Remaining Missing Values:', cleaned_df.isnull().sum().sum())
```

### Part 3: Statistical Analysis & EDA

```python
# Descriptive Statistics (Mean, Median, Mode, Std, Variance)
numeric_features = ['Age', 'Years_Experience', 'Monthly_Salary', 'Performance_Score', 'Overtime_Hours']

stats_summary = pd.DataFrame({
    'Mean': cleaned_df[numeric_features].mean(),
    'Median': cleaned_df[numeric_features].median(),
    'Mode': cleaned_df[numeric_features].mode().iloc[0],
    'Std Dev': cleaned_df[numeric_features].std(),
    'Variance': cleaned_df[numeric_features].var(),
    'Min': cleaned_df[numeric_features].min(),
    'Max': cleaned_df[numeric_features].max(),
    'Range': cleaned_df[numeric_features].max() - cleaned_df[numeric_features].min()
})
stats_summary
```
